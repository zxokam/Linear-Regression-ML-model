```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import Lin_Req_Model as lrm
```

Data Pre-Processing

```python
BostonHousing = pd.read_csv('/content/BostonHousing.csv')
```

spliting the feature and target

```python
from sklearn.preprocessing import StandardScaler
```

```python
X = BostonHousing.iloc[:, :-1].values
Y = BostonHousing.iloc[:, -1].values
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
print(X)
```

```python
print(Y)
```

spliting the dataset into training & test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size = 0.33, random_state = 2)
```

Training the Linear Regression model

```python
model = lrm.Linear_Regression(learning_rate = 0.02, no_of_iterations = 1000)
```

```python
model.fit(X_train, Y_train)
```

```python
print('weight = ', model.w[0])
print('bias = ', model.b)
```

Predict Data

```python
test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

Visualizing the predicited values & actual values

```python
plt.scatter(Y_test, test_data_prediction, color='blue', alpha=0.5)
plt.plot([Y_test.min(), Y_test.max()], [Y_test.min(), Y_test.max()], color='red', lw=2)
```

```python
from sklearn.decomposition import PCA
import seaborn as sns

# 1. Project 10D data to 2D using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 2. Create visualizations
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Plot PCA projection
scatter = axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=Y, cmap='viridis', alpha=0.8)
axes[0].set_title('10D Features Projected to 2D (PCA)')
axes[0].set_xlabel('Principal Component 1')
axes[0].set_ylabel('Principal Component 2')
fig.colorbar(scatter, ax=axes[0], label='Boston Housing (Y)')

# Plot Correlation Heatmap
sns.heatmap(BostonHousing.corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=axes[1])
axes[1].set_title('Correlation Matrix of All 11 Variables')

plt.tight_layout()
plt.show()
```
